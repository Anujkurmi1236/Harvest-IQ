```python
features_dir = Path("ml/yield_pred/data/features/")

if not features_dir.exists():
    logger.error(f"Directory does not exist: {features_dir.absolute()}")
else:
    all_files = list(features_dir.glob("*"))
    logger.info(f"Total files in directory: {len(all_files)}")
    
    npy_files = list(features_dir.glob("*.npy"))
    logger.info(f"Found {len(npy_files)} NPY files")
    
    if all_files:
        logger.info(f"First 5 files: {[f.name for f in all_files[:5]]}")
```
